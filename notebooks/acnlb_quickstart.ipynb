# acnlb quickstart

Run the alpha CN Legal Bench harness in Colab. Bring your own OpenAI-compatible
key (a Colab secret named `MODEL_API_KEY` is ideal). Nothing here is reimplemented —
this notebook calls the same `python -m acnlb` CLI the GitHub Actions workflow does.

```python
#@title Setup { display-mode: "form" }
REPO = "https://github.com/OWNER/acnlb.git"  #@param {type:"string"}
!git clone -q $REPO acnlb 2>/dev/null || (cd acnlb && git pull -q)
%cd acnlb
!pip install -q pyyaml
```

```python
#@title Provide the model key (kept in memory only) { display-mode: "form" }
import os
from getpass import getpass
try:
    from google.colab import userdata
    os.environ["MODEL_API_KEY"] = userdata.get("MODEL_API_KEY")
except Exception:
    os.environ["MODEL_API_KEY"] = getpass("MODEL_API_KEY (not written to disk): ")
os.environ.setdefault("MODEL_BASE_URL", "https://api.openai.com/v1")
print("key set:", bool(os.environ.get("MODEL_API_KEY")))
```

```python
#@title Validate the sample items and score a frozen fixture (no key needed)
!python -m acnlb validate --items items/sample
!python -m acnlb score \
  --response items/sample/fixtures/core-tx-0001.good.txt \
  --verdicts items/sample/fixtures/core-tx-0001.good.verdicts.json \
  --rubric   items/sample/core-tx-0001.rubric.md
```

```python
#@title Run the model over the sample set { display-mode: "form" }
ITEM_PATH = "items/sample"  #@param {type:"string"}
MODEL = "models.example.yaml"  #@param {type:"string"}
JUDGE = "manual"  #@param ["manual", "llm"]
!python -m acnlb run --items $ITEM_PATH --model $MODEL --judge $JUDGE --out runs/colab
```
